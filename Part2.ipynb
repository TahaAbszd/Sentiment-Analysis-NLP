## Aspect Based Sentiment Analysis

```python
import pandas as pd
import gensim.downloader as api
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
import re
```

#### Load DataSet

```python
data = pd.read_csv(r'/home/void/Desktop/Folder/QBC/NLP/Datas/train_data.csv')
```

```python
stopword = set(stopwords.words('english'))
```

#### Function to clean the Review Text

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]','',text)
    words = text.split()
    words = [w for w in words if w not in stopword]
    return words
```

#### Added Word Embedding Model to find Similar Words

```python
model = api.load("glove-wiki-gigaword-100")
```

#### Find 15 similar words to Guarantee and Warrenty

```python
warrenty_similar = model.most_similar('warranty',topn=15)
guarantee_similar = model.most_similar('guarantee',topn=15)
warsim = []
guasim = []
for i in range(15):
    name1, num1 = warrenty_similar[i]
    warsim.append(name1)
    name2, num2 = guarantee_similar[i]
    guasim.append(name2)

print(warsim)
print(guasim)
```

#### Added some typing Errors

```python
typerr = ['warrenty',
'waranty',
'garantee',
'gurantee',
'guarentee'
'garante'
'waranti']
```

```python
keyword = []
keyword.extend(typerr)
keyword.extend(guasim)
keyword.extend(warsim)
print(keyword)
```

#### Filter the dataset contains the Keywords

```python
def filterwar(text):
    for w in text:
        if w in keyword:
            return True
        return False
    
data['tokens'] = data['reviewText'].apply(clean_text)
```

```python
filtered_df = data[data['tokens'].apply(filterwar)]
filtered_df
```

#### Calculate the mean score of this Products

```python
mean_score_warranty = filtered_df.groupby('asin')['overall'].mean()
mean_score_warranty
```
